# Check Prime Number

```python
number = int(input('Enter your number here: '))

def is_prime(number):
    if number < 2:
        return False

    for i in range(2,number):
        if number % i == 0:
            return False

    return True

val = is_prime(number)
print(val)
```

# Calculate Factorial

# Check Palindrome String

```python
# A Palindrome String is same as Original String if Reversed
# For example: 1.racecar
#              2.abcddcba etc...

def is_palindrome(astring):
    astring = astring.lower()
    reverse_str = astring[::-1]
    if astring == reverse_str:
        return True
    else:
        return False

string = input("Enter your input here: ")
is_palindrome(string)
```

# Check Leap Year

```python
# Should be divisible by 4 and not by 100
# Or Should be divisible by 400

year = int(input("Enter the year: "))

def check_leapyear(year):
    if year % 4 == 0:
        if year % 100 != 0:
            print(f"{year} is a leap year")
        elif year % 100 == 0:
            if year % 400 == 0:
                print(f"{year} is a leap year")
            else:
                print(f"{year} is not a leap year")
    else:
        print(f"{year} is not a leap year")

check_leapyear(year)
        
```

# Calculate LCM

```python
import math

def calc_lcm(a,b):
    gcd = math.gcd(a,b)
    lcm = (a*b)/gcd
    return lcm

a = 12
b = 42

lcm = calc_lcm(a,b)
print(f"LCM of {a}, {b} is {lcm}")
```

# Check if Armstrong Number

```python
number = 67465
str_num = str(number)
len_num = len(str_num)

arm_strongnum = sum(int(digit) ** len_num for digit in str_num)
if number == arm_strongnum:
    print(f"{number} is an armstrong number")
else:
    print(f"{number} is not an armstrong number")
```

# Calculate Area of a Triangle

```python
import math
def calc_are(sidea, sideb, sidec):
    semi_perimeter = (sidea + sideb + sidec)/2
    area_sq = semi_perimeter*(semi_perimeter-sidea)*(semi_perimeter-sideb)*(semi_perimeter-sidec)
    return math.sqrt(area_sq)

sidea = input('Enter length of side 1:')
sideb = input('Enter length of side 2:')
sidec = input('Enter length of side 3:')

area = calc_are(int(sidea), int(sideb), int(sidec))

print(f"The area is {area}")
```
